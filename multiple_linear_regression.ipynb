# Multiple Linear Regression on real estate data(with and without dummy variables)
This notebook walks through two versions of a multiple linear regression on the same real estate problem:

1. **Part 1:** Regression using `size` and `year` only.
2. **Part 2:** Regression using `size`, `year`, and `view` (with `view` converted into a dummy variable).

Comparing the two shows how adding a well-chosen categorical variable (encoded as a dummy) can improve model fit.

## Part 1: Regression with `size` and `year`

## Import the relevant libraries

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()
```

## Load the data 

```python
data = pd.read_csv('real_estate_price_size_year.csv')
pd.set_option('display.max_rows', 10)
```

```python
data 
```

```python
data.describe()
```

## Create the regression

### Declare the dependant and independant variables

```python
y = data ['price']
x1 = data [['size', 'year']]
```

### Regression

```python
x = sm.add_constant(x1)
results = sm.OLS(y,x).fit()
results.summary()
```

## Part 2: Regression with `size`, `year`, and `view` (dummy variable)

## Load the data 

```python
more_data = pd.read_csv('real_estate_price_size_year_view.csv')
pd.set_option('display.max_rows', 10)
```

```python
more_data
```

```python
data = more_data.copy()
```

```python
data['view'] = data['view'].map({'No sea view' : 1 , 'Sea view' : 0})
```

```python
data
```

```python
data.describe(include='all')
```

## Create the regression
### Declare the dependant and independant variables

```python
p = data['price']
q1 = data[['size','year','view']]
```

## regression

```python
q = sm.add_constant(q1)
results = sm.OLS(p,q).fit()
results.summary()
```

## Comparison & Takeaway

| Model | R-squared | Adj. R-squared | Notes |
|---|---|---|---|
| Part 1: size + year | 0.776 | 0.772 | Two numeric predictors only |
| Part 2: size + year + view | 0.913 | 0.910 | Adding the `view` dummy substantially improves fit |

Adding the `view` dummy variable increased R-squared from **0.776 to 0.913**, showing that whether a property has a sea view is a meaningful, statistically significant predictor of price (coefficient ≈ $56,730, p = 0.000).
